```python
import math

# Menyimpan data diri Giant
nama = "Giant"
usia = 20
berat_badan = 97.5
tinggi_badan = 1.75
status_olahraga = False
```

```python
# Berat badan sebagai string, kemudian dikonversi menjadi float
berat_string = "97.5"
berat = float(berat_string)
print(f"Halo, saya {nama}. Usia saya {usia} tahun dan berat saya {berat} kg.")
```

```python
# Menghitung BMI
bmi = berat / (tinggi_badan ** 2)

print(f"BMI Giant adalah {bmi:.2f}")
```

```python
# Menghitung target minum air
target_air = 2000
ukuran_botol = 350

botol_penuh = target_air // ukuran_botol
sisa_air = target_air % ukuran_botol

print(f"Target minum air: {target_air} ml")
print(f"Botol penuh yang dibutuhkan: {botol_penuh} botol")
print(f"Sisa air: {sisa_air} ml")

# True division
hasil_division = target_air / ukuran_botol
print(f"Hasil true division: {hasil_division}")
```

```python
# Menghitung kebutuhan kalori
kalori = 2100.5

kalori_ke_bawah = math.floor(kalori)
kalori_ke_atas = math.ceil(kalori)

print(f"Kalori dibulatkan ke bawah: {kalori_ke_bawah}")
print(f"Kalori dibulatkan ke atas: {kalori_ke_atas}")
```
