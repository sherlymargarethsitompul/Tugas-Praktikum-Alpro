```python
# Membuat list belanjaan
belanjaan = ["Apel", "Susu", "Roti"]

# Membuat tuple harga dan jumlah
# Format: (harga1, jumlah1, harga2, jumlah2, harga3, jumlah3)
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

# Menampilkan data belanjaan
print("List Belanjaan:", belanjaan)
print("Tuple Harga dan Jumlah:", harga_jumlah)

# Menampilkan karakter pertama barang pertama
print("Karakter pertama barang pertama:", belanjaan[0][0])

# Menampilkan karakter terakhir barang terakhir
print("Karakter terakhir barang terakhir:", belanjaan[-1][-1])
```

```python
# Mengambil harga dan jumlah setiap barang
harga1, jumlah1 = harga_jumlah[0], harga_jumlah[1]
harga2, jumlah2 = harga_jumlah[2], harga_jumlah[3]
harga3, jumlah3 = harga_jumlah[4], harga_jumlah[5]

# Menghitung total harga masing-masing barang
total1 = harga1 * jumlah1
total2 = harga2 * jumlah2
total3 = harga3 * jumlah3

# Menghitung total seluruh belanja
total_belanja = total1 + total2 + total3

print("\nTotal harga Apel:", total1)
print("Total harga Susu:", total2)
print("Total harga Roti:", total3)
print("Total seluruh belanja:", total_belanja)
```

```python
# Menyimpan data pembeli dan toko
pembeli = "Nobita Nobi"
toko = "Toko Sembako Doraemon"

# Mencetak struk menggunakan f-string
print("\n" + "=" * 42)
print(f"{toko:^42}")
print("=" * 42)
print(f"Pembeli : {pembeli}")
print("-" * 42)
print(f"{'Barang':<12}{'Harga':>10}{'Jumlah':>8}{'Total':>12}")
print("-" * 42)

print(f"{belanjaan[0]:<12}{harga1:>10}{jumlah1:>8}{total1:>12}")
print(f"{belanjaan[1]:<12}{harga2:>10}{jumlah2:>8}{total2:>12}")
print(f"{belanjaan[2]:<12}{harga3:>10}{jumlah3:>8}{total3:>12}")

print("-" * 42)
print(f"{'Total Belanja':<30}{total_belanja:>12}")
print("=" * 42)
print(f"{'Terima kasih sudah belanja!':^42}")
print("=" * 42)
```
