```python
# Meminta user memasukkan nama makanan, karakter, kata sifat
```

```python
nama_makanan = input("Masukkan nama makanan: ")
nama_karakter = input("Masukkan nama karakter: ")
kata_sifat = input("Masukkan kata sifat: ")
```

```python
# Menggabungkan ketiga variabel menjadi nama toko
```

```python
nama_toko = nama_makanan + " " + nama_karakter + " " + kata_sifat
```

```python
# Menampilkan nama resmi toko
```

```python
print("Nama resmi toko:", nama_toko)
```

```python
# Menyimpan modal Doraemon, nobita dan shizuka serta menghitung total
```

```python
modal_doraemon = 555000
modal_nobita = 333000
modal_shizuka = 222000
total_modal = modal_doraemon + modal_nobita + modal_shizuka
print("Total modal: Rp", total_modal)
```

```python
# Menyimpan jumlah dorayaki yang terjual
```

```python
jumlah_dorayaki = 350
harga_dorayaki = 5000
total_pendapatan = jumlah_dorayaki * harga_dorayaki
print("Total pendapatan: Rp", total_pendapatan)
```

```python
# Menghitung keuntungan bulan pertama
```

```python
keuntungan = total_pendapatan - total_modal
print(keuntungan)
```

```python
print("keuntungan total adalah", keuntungan)
```
