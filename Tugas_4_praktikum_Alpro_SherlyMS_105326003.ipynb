```python
usia_pengemudi = int(input("Masukkan usia pengemudi (tahun): "))
jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").strip().lower()
nilai_kendaraan = float(input("Masukkan nilai kendaraan (Rp): "))
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").strip().lower()
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").strip().lower()

# #1 Kategori Risiko
if jenis_kendaraan == "motor":
    risiko_level = 2
else:
    risiko_level = 1

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko_level = 3

if area_penggunaan == "perkotaan" and risiko_level < 3:
    risiko_level += 1

if sistem_pengaman == "tidak":
    if risiko_level < 3:
        risiko_level += 1
elif sistem_pengaman == "ya":
    pass
else:
    pass

if risiko_level == 1:
    kategori_risiko = "rendah"
elif risiko_level == 2:
    kategori_risiko = "sedang"
else:
    kategori_risiko = "tinggi"

# #2 Premi Asuransi yang Perlu Dibayar
if kategori_risiko == "tinggi":
    persentase_premi = 0.1
elif kategori_risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi_asuransi = persentase_premi * nilai_kendaraan

# #3 Pesan Doraemon
pesan = (
    "Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if kategori_risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon"
)

# Cetak hasil
print("\nHASIL ANALISIS:")
print("1. Kategori risiko:", kategori_risiko)
print("2. Premi asuransi yang perlu dibayarkan: Rp", premi_asuransi)
print("3. Pesan Doraemon:", pesan)
```
